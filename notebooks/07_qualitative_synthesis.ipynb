# Qualitative Synthesis & Argument Construction

```python
import pandas as pd
import numpy as np
```

## 1. Load the Final Curated Reading List

```python
final_list_df = pd.read_csv('./data/processed/final_reading_list.csv')

print(f"Loaded {len(final_list_df)} papers for deep reading and qualitative analysis.")
```

## 2. Define the Full Qualititative Analysis Framework

```python
qualitative_columns = [
    # Identification Columns
    'paperId', 'Title', 'Year', 'Primary_Pillar(s)', 'Methodology_Group', 'Priority_Score',
    
    # Core Synthesis Questions
    'Core_Problem', 'Key_Contribution', 'Approach_to_Grid_Stability', 
    'Approach_to_Trust_Explainability', 'Stated_Limitations', 'Future_Work_Research_Gap',
    
    # Methodology-Specific Questions
    'Causal_Elements', 'Technical_Innovation',
    'Key_Trends_Identified', 'Proposed_Taxonomy',
    'Context_Data', 'Key_Metrics',
    
    # Researcher's Synthesis
    'Researcher_Synthesis_Connections'
]

# Create the new, empty DataFrame
analysis_df = pd.DataFrame(columns=qualitative_columns)
```

## 3. Populate the Framework

```python
analysis_df['paperId'] = final_list_df['paperId']
analysis_df['Title'] = final_list_df['title']
analysis_df['Year'] = final_list_df['year']
analysis_df['Methodology_Group'] = final_list_df['methodology_group']
analysis_df['Priority_Score'] = final_list_df['priority_score']
analysis_df['Key_Contribution'] = final_list_df['contribution']
```

## 4. Programmatically Determine the Primary Pillar(s)

```python
def get_primary_pillars(row):
    scores = {
        'CRL': row['score_crl_methods'],
        'EV': row['score_ev_applications'],
        'XAI': row['score_explainability_trust']
    }
    # Find the maximum score
    max_score = max(scores.values())
    # A paper is considered a primary pillar if its score is the max score AND the score is high (e.g., >=7)
    primary = [pillar for pillar, score in scores.items() if score == max_score and score >= 7]
    if not primary:
        # If no score is >= 7, just take the one with the highest score
        primary = [max(scores, key=scores.get)]
    return ' & '.join(primary)

analysis_df['Primary_Pillar(s)'] = final_list_df.apply(get_primary_pillars, axis=1)
```

## 5. Save the Framework to a CSV File

```python
analysis_df.to_csv('./data/processed/qualitative_analysis_sheet.csv', index=False)

print(f"\nSuccessfully created 'qualitative_analysis_sheet.csv' with {len(analysis_df)} papers.")
print("The sheet is ready for manual data entry and qualitative synthesis.")

# Display the head of our new analysis framework to verify.
display(analysis_df.head())
```
